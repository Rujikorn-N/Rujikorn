```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# 0) Load and prepare data
albumin = pd.read_csv("albumin_100k.csv")
f8 = pd.read_csv("f8_100k.csv")

# Add label
albumin["label"] = 0   # HSA
f8["label"] = 1        # F8

# Combined data
raw_df = pd.concat([albumin, f8], ignore_index=True)

# Cleaning data
prep_df = raw_df.dropna().copy()

if "ignored" in prep_df.columns:
    prep_df = prep_df.loc[prep_df["ignored"] == False].copy()

# Add Features
feature_list = ["dwell_time_ms", "amplitude", "amplitude_sd"]
prep_df = prep_df[feature_list + ["label"]].copy()

print("Prepared dataset shape:", prep_df.shape)
print(prep_df["label"].value_counts())

# Save fig
base_save_dir = "model evaluation 100 kHz"
os.makedirs(base_save_dir, exist_ok=True)

# Train / Test split
X = prep_df[feature_list].copy()
y = prep_df["label"].copy()
orig_idx = prep_df.index.copy()  

X_train, X_test, y_train, y_test, idx_train, idx_test = train_test_split(
    X, y, orig_idx,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("\nTrain shape:", X_train.shape)
print("Test shape :", X_test.shape)

# Model Training and Validation (Model tuning)
param_grids_original = {
    "Decision Tree": {
        "pipeline": Pipeline([
            ("scaler", StandardScaler()),
            ("model", DecisionTreeClassifier(random_state=42))
        ]),
        "params": {
            "model__max_depth": [3, 5, 7],
            "model__criterion": ["gini", "entropy"],
            "model__class_weight": ["balanced"]
        }
    },
    "Random Forest": {
        "pipeline": Pipeline([
            ("scaler", StandardScaler()),
            ("model", RandomForestClassifier(random_state=42, n_jobs=-1))
        ]),
        "params": {
            "model__n_estimators": [10, 50, 100, 200],
            "model__max_depth": [2, 5, 10, 15],
            "model__class_weight": ["balanced"]
        }
    },
    "Nearest Neighbor": {
        "pipeline": Pipeline([
            ("scaler", StandardScaler()),
            ("model", KNeighborsClassifier())
        ]),
        "params": {
            "model__n_neighbors": [5, 7, 9],
            "model__weights": ["uniform", "distance"]
        }
    }
}

param_grids_pca_only = {
    "Decision Tree": {
        "pipeline": Pipeline([
            ("scaler", StandardScaler()),
            ("pca", PCA(n_components=3)),
            ("model", DecisionTreeClassifier(random_state=42))
        ]),
        "params": {
            "model__max_depth": [3, 5, 7],
            "model__criterion": ["gini", "entropy"],
            "model__class_weight": ["balanced"]
        }
    },
    "Random Forest": {
        "pipeline": Pipeline([
            ("scaler", StandardScaler()),
            ("pca", PCA(n_components=3)),
            ("model", RandomForestClassifier(random_state=42, n_jobs=-1))
        ]),
        "params": {
            "model__n_estimators": [10, 50, 100, 200],
            "model__max_depth": [2, 5, 10, 15],
            "model__class_weight": ["balanced"]
        }
    },
    "Nearest Neighbor": {
        "pipeline": Pipeline([
            ("scaler", StandardScaler()),
            ("pca", PCA(n_components=3)),
            ("model", KNeighborsClassifier())
        ]),
        "params": {
            "model__n_neighbors": [5, 7, 9],
            "model__weights": ["uniform", "distance"]
        }
    }
}

# 5-fold cross validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

def plot_confusion_matrix_percent(y_true, y_pred, title, experiment_name, model_name):
    cm = confusion_matrix(y_true, y_pred)
    cm_perc = cm.astype(float) / cm.sum(axis=1, keepdims=True) * 100

    labels = [f"{v:.1f}%" for v in cm_perc.flatten()]
    labels = np.array(labels).reshape(cm.shape)

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm_perc,
        annot=labels,
        fmt="",
        cmap="Blues",
        cbar_kws={"label": "Recall (%)"},
        xticklabels=["HSA", "F8"],
        yticklabels=["HSA", "F8"]
    )
    plt.title(title, fontsize = 15)
    plt.xlabel("Predicted Label", fontsize = 12)
    plt.ylabel("Actual Label", fontsize = 12)

    plt.tight_layout()

    safe_exp = re.sub(r'[^\w\-_. ]', '_', experiment_name)
    save_path = os.path.join(base_save_dir, safe_exp)
    os.makedirs(save_path, exist_ok=True)

    safe_model = re.sub(r'[^\w\-_. ]', '_', model_name)
    filename = f"{safe_model}.png"

    plt.savefig(os.path.join(save_path, filename), dpi=300, bbox_inches="tight")
    plt.show()


def plot_misclassified_points(X_test_df, y_test_ser, mis_only, model_name, save_path):
    plt.figure(figsize=(7, 5))

    # จุดทั้งหมดใน test data
    plt.scatter(
        X_test_df["dwell_time_ms"],
        X_test_df["amplitude"],
        c=y_test_ser,
        cmap="coolwarm",
        alpha=0.30,
        s=35,
        label="All test samples"
    )

    # จุดที่ทำนายพลาด
    if len(mis_only) > 0:
        plt.scatter(
            mis_only["dwell_time_ms"],
            mis_only["amplitude"],
            color="yellow",
            edgecolor="black",
            s=90,
            label="Misclassified"
        )

    plt.xlabel("dwell_time_ms")
    plt.ylabel("amplitude")
    plt.title(f"Misclassified Points: {model_name}")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(os.path.join(save_path, f"{model_name}_misclassified_plot.png"), dpi=300, bbox_inches="tight")
    plt.show()


def analyze_misclassification(X_test_df, y_test_ser, y_pred, idx_test_ser, experiment_name, model_name):
    label_map = {0: "HSA", 1: "F8"}

    mis_df = X_test_df.copy()
    mis_df["actual_label"] = y_test_ser.values
    mis_df["pred_label"] = y_pred
    mis_df["orig_index"] = idx_test_ser.values if hasattr(idx_test_ser, "values") else idx_test_ser
    mis_df["actual_name"] = mis_df["actual_label"].map(label_map)
    mis_df["pred_name"] = mis_df["pred_label"].map(label_map)

    mis_only = mis_df[mis_df["actual_label"] != mis_df["pred_label"]].copy()

    hsa_to_f8 = mis_only[
        (mis_only["actual_label"] == 0) & (mis_only["pred_label"] == 1)
    ].copy()

    f8_to_hsa = mis_only[
        (mis_only["actual_label"] == 1) & (mis_only["pred_label"] == 0)
    ].copy()

    safe_exp = re.sub(r'[^\w\-_. ]', '_', experiment_name)
    safe_model = re.sub(r'[^\w\-_. ]', '_', model_name)
    save_path = os.path.join(base_save_dir, safe_exp)
    os.makedirs(save_path, exist_ok=True)

    mis_only.to_csv(os.path.join(save_path, f"{safe_model}_misclassified.csv"), index=False)
    hsa_to_f8.to_csv(os.path.join(save_path, f"{safe_model}_HSA_to_F8.csv"), index=False)
    f8_to_hsa.to_csv(os.path.join(save_path, f"{safe_model}_F8_to_HSA.csv"), index=False)

    print(f"\n Misclassified samples ({model_name}): {len(mis_only)}")
    if len(mis_only) > 0:
        print(mis_only[[
            "orig_index", "dwell_time_ms", "amplitude", "amplitude_sd",
            "actual_name", "pred_name"
        ]].head(20))
    else:
        print("No misclassified samples.")

    print(f"\n HSA -> F8 errors: {len(hsa_to_f8)}")
    if len(hsa_to_f8) > 0:
        print(hsa_to_f8[[
            "orig_index", "dwell_time_ms", "amplitude", "amplitude_sd",
            "actual_name", "pred_name"
        ]].head(10))

    print(f"\n F8 -> HSA errors: {len(f8_to_hsa)}")
    if len(f8_to_hsa) > 0:
        print(f8_to_hsa[[
            "orig_index", "dwell_time_ms", "amplitude", "amplitude_sd",
            "actual_name", "pred_name"
        ]].head(10))

    plot_misclassified_points(X_test_df, y_test_ser, mis_only, safe_model, save_path)

    return mis_only, hsa_to_f8, f8_to_hsa

# Run Experiment
def run_grid_search_experiment(experiment_name, param_grids, X_train, X_test, y_train, y_test, idx_test):
    print("=" * 70)
    print(f"EXPERIMENT: {experiment_name}")
    print("=" * 70)

    best_models = {}
    results_summary = {}

    for name, config in param_grids.items():
        print(f"\nSearching best params for: {name}")

        grid = GridSearchCV(
            estimator=config["pipeline"],
            param_grid=config["params"],
            cv=skf,
            scoring="f1_macro",
            n_jobs=-1,
            verbose=0,
            refit=True
        )
        grid.fit(X_train, y_train)

        best_models[name] = grid.best_estimator_

        print("Best Params:", grid.best_params_)
        print(f"Best CV F1-macro: {grid.best_score_:.4f}")

        y_pred = grid.best_estimator_.predict(X_test)
        report = classification_report(y_test, y_pred, output_dict=True)

        test_f1 = report["macro avg"]["f1-score"]
        f1_weighted = report["weighted avg"]["f1-score"]

        print(f"\n--- Test Result: {name} ---")
        print(classification_report(y_test, y_pred, target_names=["HSA", "F8"]))
        print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
        print(f"F1-macro   : {test_f1:.4f}")
        print(f"F1-weighted: {f1_weighted:.4f}")

        plot_confusion_matrix_percent(
            y_test,
            y_pred,
            title=f"{experiment_name}\n{name} | F-value = {test_f1*100:.1f}%",
            experiment_name=experiment_name,
            model_name=name
        )

        # analyze misclassification
        mis_only, hsa_to_f8, f8_to_hsa = analyze_misclassification(
            X_test_df=X_test,
            y_test_ser=y_test,
            y_pred=y_pred,
            idx_test_ser=idx_test,
            experiment_name=experiment_name,
            model_name=name
        )

        results_summary[name] = {
            "Experiment": experiment_name,
            "Model": name,
            "Best_CV_F1_macro": grid.best_score_,
            "Test_F1_macro": test_f1,
            "Misclassified": len(mis_only),
            "HSA_to_F8": len(hsa_to_f8),
            "F8_to_HSA": len(f8_to_hsa)
        }

    results_df = pd.DataFrame(results_summary.values()).sort_values(
        by="Test_F1_macro", ascending=False
    ).reset_index(drop=True)

    print("\nSummary:")
    print(results_df)

    return best_models, results_df


def load_and_prepare_mixture(file_path, feature_cols):
    df = pd.read_csv(file_path)
    df = df.dropna().copy()

    if "ignored" in df.columns:
        df = df.loc[df["ignored"] == False].copy()

    X_mix = df[feature_cols].copy()
    return df, X_mix


def predict_mixture_with_pipeline(file_path, mixture_name, fitted_pipeline, feature_cols):
    df_mix, X_mix = load_and_prepare_mixture(file_path, feature_cols)

    preds = fitted_pipeline.predict(X_mix)
    df_mix["predicted_label"] = preds
    df_mix["predicted_name"] = df_mix["predicted_label"].map({0: "HSA", 1: "F8"})

    total = len(df_mix)
    count_hsa = (df_mix["predicted_label"] == 0).sum()
    count_f8 = (df_mix["predicted_label"] == 1).sum()

    print("\n" + "-" * 50)
    print(f"{mixture_name}")
    print("-" * 50)
    print(f"Total events : {total}")
    print(f"HSA predicted: {count_hsa} ({count_hsa/total*100:.1f}%)")
    print(f"F8 predicted : {count_f8} ({count_f8/total*100:.1f}%)")

    return df_mix

# Experiment 1: Original 3 features
best_models_original, summary_original = run_grid_search_experiment(
    experiment_name="Original 3 Features",
    param_grids=param_grids_original,
    X_train=X_train,
    X_test=X_test,
    y_train=y_train,
    y_test=y_test,
    idx_test=idx_test
)

best_original_name = summary_original.iloc[0]["Model"]
best_original_model = best_models_original[best_original_name]

print(f"\nBest model for Original 3 Features: {best_original_name}")

mix1_original = predict_mixture_with_pipeline(
    "mixture_1.csv", "Mixture 1 | Original 3 Features",
    best_original_model, feature_list
)

mix2_original = predict_mixture_with_pipeline(
    "mixture_2.csv", "Mixture 2 | Original 3 Features",
    best_original_model, feature_list
)

# Experiment 2: PCA-only (3 components)
best_models_pca, summary_pca = run_grid_search_experiment(
    experiment_name="PCA Only (3 Components)",
    param_grids=param_grids_pca_only,
    X_train=X_train,
    X_test=X_test,
    y_train=y_train,
    y_test=y_test,
    idx_test=idx_test
)

best_pca_name = summary_pca.iloc[0]["Model"]
best_pca_model = best_models_pca[best_pca_name]

print(f"\nBest model for PCA Only: {best_pca_name}")

mix1_pca = predict_mixture_with_pipeline(
    "mixture_1.csv", "Mixture 1 | PCA Only",
    best_pca_model, feature_list
)

mix2_pca = predict_mixture_with_pipeline(
    "mixture_2.csv", "Mixture 2 | PCA Only",
    best_pca_model, feature_list
)

# Compare all experiments
all_summary = pd.concat(
    [summary_original, summary_pca],
    ignore_index=True
).sort_values(by="Test_F1_macro", ascending=False).reset_index(drop=True)

print("\n" + "=" * 70)
print("FINAL COMPARISON")
print("=" * 70)
print(all_summary)
```

```python
# Threshold-based mixture prediction
def predict_mixture_with_threshold(file_path, mixture_name, model, feature_cols, threshold=0.5):
    df_mix = pd.read_csv(file_path).dropna().copy()

    if "ignored" in df_mix.columns:
        df_mix = df_mix.loc[df_mix["ignored"] == False].copy()

    X_mix = df_mix[feature_cols].copy()

    proba_f8 = model.predict_proba(X_mix)[:, 1]
    pred = (proba_f8 >= threshold).astype(int)

    df_mix["prob_F8"] = proba_f8
    df_mix["predicted_label"] = pred
    df_mix["predicted_name"] = df_mix["predicted_label"].map({0: "HSA", 1: "F8"})

    total = len(df_mix)
    count_hsa = (df_mix["predicted_label"] == 0).sum()
    count_f8 = (df_mix["predicted_label"] == 1).sum()

    print("\n" + "-" * 55)
    print(f"{mixture_name} | decision threshold = {threshold}")
    print("-" * 55)
    print(f"Total events : {total}")
    print(f"HSA predicted: {count_hsa} ({count_hsa/total*100:.1f}%)")
    print(f"F8 predicted : {count_f8} ({count_f8/total*100:.1f}%)")

    return df_mix


def predict_mixture_with_confidence(file_path, mixture_name, model, feature_cols, conf_th=0.8):
    df_mix = pd.read_csv(file_path).dropna().copy()

    if "ignored" in df_mix.columns:
        df_mix = df_mix.loc[df_mix["ignored"] == False].copy()

    X_mix = df_mix[feature_cols].copy()

    proba = model.predict_proba(X_mix)
    max_proba = np.max(proba, axis=1)
    pred = np.argmax(proba, axis=1)

    pred[max_proba < conf_th] = -1

    df_mix["max_probability"] = max_proba
    df_mix["predicted_label"] = pred
    df_mix["predicted_name"] = df_mix["predicted_label"].map({
        0: "HSA",
        1: "F8",
        -1: "Uncertain"
    })

    total = len(df_mix)
    count_hsa = (df_mix["predicted_label"] == 0).sum()
    count_f8 = (df_mix["predicted_label"] == 1).sum()
    count_uncertain = (df_mix["predicted_label"] == -1).sum()

    print("\n" + "-" * 55)
    print(f"{mixture_name} | confidence threshold = {conf_th}")
    print("-" * 55)
    print(f"Total events   : {total}")
    print(f"HSA predicted  : {count_hsa} ({count_hsa/total*100:.1f}%)")
    print(f"F8 predicted   : {count_f8} ({count_f8/total*100:.1f}%)")
    print(f"Uncertain      : {count_uncertain} ({count_uncertain/total*100:.1f}%)")

    return df_mix
    

best_model_for_mixture = best_original_model

# ---------- decision threshold ----------
mix1_th_05 = predict_mixture_with_threshold(
    "mixture_1.csv",
    "Mixture 1",
    best_model_for_mixture,
    feature_list,
    threshold=0.5
)

mix2_th_05 = predict_mixture_with_threshold(
    "mixture_2.csv",
    "Mixture 2",
    best_model_for_mixture,
    feature_list,
    threshold=0.5
)

# ---------- confidence threshold ----------
mix1_conf_08 = predict_mixture_with_confidence(
    "mixture_1.csv",
    "Mixture 1",
    best_model_for_mixture,
    feature_list,
    conf_th=0.8
)

mix2_conf_08 = predict_mixture_with_confidence(
    "mixture_2.csv",
    "Mixture 2",
    best_model_for_mixture,
    feature_list,
    conf_th=0.8
)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare plot data
train_plot_df = X_train.copy()
train_plot_df["label"] = y_train.values
train_plot_df["label_name"] = train_plot_df["label"].map({0: "HSA", 1: "F8"})
train_plot_df["log_dwell_time"] = np.log10(train_plot_df["dwell_time_ms"])

mix1_plot_df = mix1_original.copy()
mix2_plot_df = mix2_original.copy()

mix1_plot_df["log_dwell_time"] = np.log10(mix1_plot_df["dwell_time_ms"])
mix2_plot_df["log_dwell_time"] = np.log10(mix2_plot_df["dwell_time_ms"])

# Save directory
save_dir = "EDA_plots_distribution"
os.makedirs(save_dir, exist_ok=True)

def plot_training_distributions(df, features, save_dir, prefix="train"):
    for feature in features:
        plt.figure(figsize=(7, 5))

        sns.histplot(
            data=df,
            x=feature,
            hue="label_name",
            bins=50,
            stat="density",
            common_norm=False,
            kde=True
        )

        plt.title(f"Training Data Distribution: {feature}", fontsize=13)
        plt.xlabel(feature)
        plt.ylabel("Density")
        plt.tight_layout()

        save_path = os.path.join(save_dir, f"{prefix}_{feature}.png")
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
        plt.show()

        print(f"Saved: {save_path}")


def plot_mixture_distributions(df, features, mixture_name, save_dir, prefix="mixture"):
    for feature in features:
        plt.figure(figsize=(7, 5))

        sns.histplot(
            data=df,
            x=feature,
            hue="predicted_name",
            bins=50,
            stat="density",
            common_norm=False,
            kde=True
        )

        plt.title(f"{mixture_name} Distribution: {feature}", fontsize=13)
        plt.xlabel(feature)
        plt.ylabel("Density")
        plt.tight_layout()

        safe_name = mixture_name.lower().replace(" ", "_")
        save_path = os.path.join(save_dir, f"{prefix}_{safe_name}_{feature}.png")
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
        plt.show()

        print(f"Saved: {save_path}")


def plot_train_vs_mixture(train_df, mix_df, feature, mixture_name, save_dir, use_predicted=True):
    plt.figure(figsize=(8, 5))

    color_map = {
        "HSA": "red",
        "F8": "blue"
    }

    mix_color_map = {
        "HSA": "orange",
        "F8": "cyan"
    }

    # Train
    sns.kdeplot(
        data=train_df[train_df["label_name"] == "HSA"],
        x=feature,
        label="Train HSA",
        linewidth=2,
        color=color_map["HSA"]
    )

    sns.kdeplot(
        data=train_df[train_df["label_name"] == "F8"],
        x=feature,
        label="Train F8",
        linewidth=2,
        color=color_map["F8"]
    )

    # Mixture
    hue_col = "predicted_name" if use_predicted else None

    if hue_col is not None:
        for label_name in ["HSA", "F8"]:   
            subset = mix_df[mix_df[hue_col] == label_name]
            if len(subset) == 0:
                continue

            sns.kdeplot(
                data=subset,
                x=feature,
                label=f"{mixture_name} Pred {label_name}",
                linestyle="--",
                linewidth=2,
                color=mix_color_map[label_name]
            )

    else:
        sns.kdeplot(
            data=mix_df,
            x=feature,
            label=f"{mixture_name} All",
            linestyle="--",
            linewidth=2,
            color="black"
        )

    plt.title(f"Train vs {mixture_name}: {feature}", fontsize=13)
    plt.xlabel(feature)
    plt.ylabel("Density")
    plt.legend()
    plt.tight_layout()

    safe_name = mixture_name.lower().replace(" ", "_")
    save_path = os.path.join(save_dir, f"compare_{safe_name}_{feature}.png")
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Saved: {save_path}")

# Plot raw features
raw_features = ["dwell_time_ms", "amplitude", "amplitude_sd"]

plot_training_distributions(train_plot_df, raw_features, save_dir, prefix="train_raw")

plot_mixture_distributions(mix1_plot_df, raw_features, "Mixture 1", save_dir, prefix="mix1_raw")
plot_mixture_distributions(mix2_plot_df, raw_features, "Mixture 2", save_dir, prefix="mix2_raw")

for feat in raw_features:
    plot_train_vs_mixture(train_plot_df, mix1_plot_df, feat, "Mixture 1", save_dir)
    plot_train_vs_mixture(train_plot_df, mix2_plot_df, feat, "Mixture 2", save_dir)

# Plot log dwell
log_features = ["log_dwell_time", "amplitude", "amplitude_sd"]

plot_training_distributions(train_plot_df, log_features, save_dir, prefix="train_log")

plot_mixture_distributions(mix1_plot_df, log_features, "Mixture 1", save_dir, prefix="mix1_log")
plot_mixture_distributions(mix2_plot_df, log_features, "Mixture 2", save_dir, prefix="mix2_log")

for feat in log_features:
    plot_train_vs_mixture(train_plot_df, mix1_plot_df, feat, "Mixture 1", save_dir)
    plot_train_vs_mixture(train_plot_df, mix2_plot_df, feat, "Mixture 2", save_dir)

print(f"\nAll plots saved in: {save_dir}")
```
